```r
library(tidyverse)
```

# Summarise Single Variable

```r
summary(mtcars$mpg)
```

```r
mtcars %>%
    summarise(mean_mpg = mean(mpg))
mtcars %>%
    summarize(mean_mpg = mean(mpg))
```

```r
mtcars %>%
    summarise(count = n())
```

# Summarise Multiple Variable

```r
df <- mtcars
```

```r
df %>%
    summarize(mean_wt = mean(wt), mean_hp = mean(hp))
```

```r
mtcars %>%
    summarize(mean_wt = mean(wt), 
              mean_hp = mean(hp), 
              hp_per_wt = mean_hp / mean_wt)
```

### summarize_all()

```r
df %>%
    summarize_all(mean)
```

```r
df %>%
    summarize_all(min)
```

### summarise_at()

```r
df %>%
    summarise_at(vars(mpg), funs(mean, median))
```

```r
df <- as_tibble(df)
df %>%
    summarise_at(vars(mpg), funs(mean, median))
```

```r
df %>%
    summarise_at(vars(mpg), funs(mean, median, max))
```

```r
df %>%
    summarise_at(vars(mpg, hp), funs(mean, median, max))
```

# group_by verb

```r
df <- mtcars
head(df)
```

```r
df %>%
    group_by(cyl) %>%
    summarise(mean_hp = mean(hp))
```

```r
df %>%
    group_by(cyl) %>%
    summarise(mean_hp = mean(hp),
             mean_wt = mean(wt),
             hp_per_wt = mean_hp / mean_wt,
             count = n())
```

```r
df %>%
    group_by(cyl, gear) %>%
    summarise(mean_hp = mean(hp),
             mean_wt = mean(wt))
```

```r
df %>%
    group_by(cyl, gear) %>%
    summarise(mean_hp = mean(hp),
             mean_wt = mean(wt),
             count = n())
```

# ungroup()

```r
df %>%
    group_by(cyl, gear) %>%
    summarise(mean_hp = mean(hp),
             mean_wt = mean(wt),
             count = n()) %>%
    ungroup()
```
