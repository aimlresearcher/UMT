```r
library(tidyverse)
```

# Creating Basic 1-D Plot

```r
df <- mtcars
head(df)
```

```r
df %>%
    ggplot(aes (x = mpg)) +
    geom_histogram()
```

```r
df %>%
    ggplot(aes (x = mpg)) +
    geom_histogram(bins = 10)
```

```r
df %>%
    ggplot(aes (x = mpg)) +
    geom_density()
```

```r
df %>%
    ggplot(aes (x = mpg)) +
    geom_density(color = 'red')
```

```r
df %>%
    ggplot(aes (x = mpg)) +
    geom_density(color = 'red', fill = 'gray')
```

# Creating Basic 2-D Plots

```r
df %>%
    ggplot(aes (x = mpg, y = wt)) +
    geom_point()
```

```r
df %>%
    ggplot(aes (x = mpg, y = wt)) +
    geom_point(color = 'purple')
```

```r
df %>%
    ggplot(aes (x = mpg, y = wt)) +
    geom_point(color = 'purple', size = 10)
```

```r
df %>%
    ggplot(aes (x = mpg, y = wt)) +
    geom_smooth()
```

```r
df %>%
    ggplot(aes (x = mpg, y = wt)) +
    geom_line()
```

```r
df %>%
    ggplot(aes (x = mpg, y = wt)) +
    geom_smooth(se = FALSE)
```

# Add Layers and Elements

```r
df <- mtcars
```

```r
df %>%
    ggplot(aes(x = mpg, y = wt)) +
    geom_point()
```

```r
df %>%
    ggplot(aes(x = mpg, y = wt)) +
    geom_point() +
    geom_smooth()
```

```r
df %>%
    ggplot(aes(x = mpg, y = wt)) +
    geom_point() +
    geom_smooth() +
    theme_bw()
```

# Visualize Summarized Data

```r
df %>% 
    ggplot(aes(x = mpg, y = wt, color = cyl)) +
    geom_point()
```

```r
df %>% 
    ggplot(aes(x = mpg, y = wt, color = as.factor(cyl))) +
    geom_point()
```

```r
df %>% 
    filter(cyl != 4) %>%
    ggplot(aes(x = mpg, y = wt)) +
    geom_point()
```
