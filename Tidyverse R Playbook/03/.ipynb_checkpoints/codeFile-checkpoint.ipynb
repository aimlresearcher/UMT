# Gather

```r
library(tidyverse)
```

```r
ldeaths
```

```r
df <- data.frame(ldeaths)
head(df)
```

```r
messy <- data.frame(name = c("hamed", "moon", "raza", "danish", "basit"),
                   a = c(99,98,78,68,96),
                   b = c(91,90,94,89,77))
messy
```

```r
messy %>%
    gather(drug, volume, a:b)
```

```r
messy <- data.frame(name = c("hamed", "moon", "raza", "danish", "basit"),
                   a = c(99,98,78,68,96),
                   b = c(91,90,94,89,77),
                   c = c(91,90,94,89,77))
messy
messy %>%
    gather(drug, volume, a:c)
```

# Spread

```r
messy
long_df <- messy %>%
    gather(drug, volume, a:c)
long_df
```

```r
long_df %>%
    spread(key = drug, value = volume)
```

```r
long_df %>%
    spread(drug, volume)
```

```r
long_df
length(long_df$name)
long_df <- long_df[-c(1, 6), ]
long_df
length(long_df$name)
```

```r
long_df %>%
spread(drug, volume)
```

```r
long_df %>%
spread(drug, volume, fill = 0)
```

# Separate and Extract
### Separate
Takes string values in a single column and splits into multiple columns
### Extract
Removes value from string column that matches specified conditions

```r
messy <- data.frame(name = c("hamed raza", "moon lahori", "raza raza", "danish kaimka", "basit kashmari"),
                   a = c(99,98,78,68,96),
                   b = c(91,90,94,89,77),
                   c = c(91,90,94,89,77))
messy
```

```r
messy %>%
separate(name, into = c("first_name", "last_name"), sep = " ")
```

```r
messy %>%
separate(name, into = c("first_name", "last_name"))
```

```r
messy %>%
separate(name, into = c("first_name", "last_name"))
```

```r
messy %>%
    extract(name, "raza", regex = "(raza)")
```

```r
messy %>%
    extract(name, "raza", regex = "(raza)", remove = FALSE)
```
